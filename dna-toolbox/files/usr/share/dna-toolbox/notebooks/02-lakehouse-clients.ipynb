# Lakehouse clients

Talks to the stack started with `dna-stack` on the host. Ports are the usual ones plus `DNA_PORT_OFFSET` (default 10000), bound to 127.0.0.1. Start it first: `dna-stack setup`, then `dna-stack up`.

```python
import os
off = int(os.environ.get('DNA_PORT_OFFSET', '10000'))
ports = {'postgres': 5432 + off, 'seaweedfs_s3': 8333 + off, 'lakekeeper': 8181 + off, 'trino': 8080 + off}
ports
```

```python
import urllib.request
for name, port in ports.items():
    try:
        urllib.request.urlopen(f'http://127.0.0.1:{port}/', timeout=2)
        print(name, 'answers')
    except Exception as e:
        print(name, 'no answer:', type(e).__name__)
```
