# dna-toolbox: welcome

This box mirrors Databricks Runtime LTS versions for local work. The pins come from `/usr/share/dna-toolbox/versions.env` (override them with the `DNA_*` variables). Photon and Unity Catalog are Databricks-only and cannot be reproduced here.

If the kernel name above does not exist, run `dna-setup` in a terminal first.

```python
import os, sys
for k in sorted(k for k in os.environ if k.startswith('DNA_') and k.endswith('_VERSION')):
    print(k, '=', os.environ[k])
print('python', sys.version.split()[0])
```
